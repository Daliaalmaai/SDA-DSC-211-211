<p>شغّل الخلية المشتركة مرة واحدة. لا تحتاج إلى بيئة مدفوعة أو GPU أو ربط Drive أو مفتاح API.</p>

```python
from pathlib import Path
import importlib.util
import os
import platform
import shutil
import sys

IN_COLAB = importlib.util.find_spec("google.colab") is not None
WORKSPACE = Path("/content/tamweel-readiness") if IN_COLAB else Path.cwd() / ".readiness-workspace"
WORKSPACE.mkdir(parents=True, exist_ok=True)
probe = WORKSPACE / "write-test.txt"
probe.write_text("ready", encoding="utf-8")
write_ok = probe.read_text(encoding="utf-8") == "ready"
probe.unlink(missing_ok=True)
disk = shutil.disk_usage(WORKSPACE)

runtime = {
    "environment": "Google Colab" if IN_COLAB else "Local/Jupyter",
    "python": platform.python_version(),
    "system": platform.system(),
    "architecture": platform.machine() or "unknown",
    "workspace_writable": write_ok,
    "free_disk_gb": round(disk.free / 1024**3, 2),
    "cpu_threads_detected": os.cpu_count(),
}
runtime
```

<td width="50%" valign="top" dir="ltr">
<h2>2. Check required packages</h2>
<p>The check does not install or upgrade anything. It identifies what is already available and gives one repair command when a package is missing.</p>
</td>
<td width="50%" valign="top" dir="rtl">
<h2>2. افحص الحزم المطلوبة</h2>
<p>لا يثبت الفحص أي حزمة ولا يرقّيها. يحدد المتاح حاليًا ويعرض أمر إصلاح واحدًا عند فقدان حزمة.</p>
</td>
</tr>
</table>

```python
from importlib import metadata

PACKAGE_IMPORTS = {
    "numpy": "numpy",
    "pandas": "pandas",
    "scikit-learn": "sklearn",
    "matplotlib": "matplotlib",
    "xgboost": "xgboost",
    "lightgbm": "lightgbm",
    "shap": "shap",
    "optuna": "optuna",
}
rows = []
missing = []
for distribution, module in PACKAGE_IMPORTS.items():
    available = importlib.util.find_spec(module) is not None
    version = None
    if available:
        try:
            version = metadata.version(distribution)
        except metadata.PackageNotFoundError:
            version = "available"
    else:
        missing.append(distribution)
    rows.append({"package": distribution, "status": "PASS" if available else "INSTALL", "version": version})

try:
    import pandas as pd
    package_table = pd.DataFrame(rows)
    display(package_table)
except Exception:
    package_table = rows
    print(*rows, sep="\n")

if missing:
    print("Missing packages | حزم غير متاحة:", ", ".join(missing))
    print("Repair command | أمر الإصلاح:")
    print(f"{sys.executable} -m pip install " + " ".join(missing))
else:
    print("Required packages detected | تم العثور على الحزم المطلوبة")
```

</td>
<td width="50%" valign="top" dir="rtl">
<h2>3. اقرأ النتيجة</h2>
<ul><li><strong>READY:</strong> مساحة العمل قابلة للكتابة وPython مدعوم والحزم المطلوبة متاحة.</li><li><strong>REVIEW:</strong> يمكن متابعة الدفتر، لكن يلزم فحص مورد أو إصدار.</li><li><strong>INSTALL:</strong> شغّل أمر الإصلاح الظاهر، ثم أعد تشغيل الجلسة وشغّل جميع الخلايا من البداية.</li></ul>

```python
python_supported = sys.version_info >= (3, 10)
all_packages = not missing
enough_disk = runtime["free_disk_gb"] >= 1.0
status = "READY" if python_supported and write_ok and all_packages and enough_disk else "REVIEW"
readiness = {
    "status": status,
    "python_supported": python_supported,
    "workspace_writable": write_ok,
    "required_packages_detected": all_packages,
    "minimum_free_disk_available": enough_disk,
    "paid_subscription_required": False,
    "api_key_required": False,
    "gpu_required": False,
}
print(f"{status} | {'جاهز للبدء' if status == 'READY' else 'راجع البنود غير المكتملة'}")
readiness
```
